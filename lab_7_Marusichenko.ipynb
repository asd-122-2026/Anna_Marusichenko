## **Завдання 1**: Створити бінарне дерево згідно з варіантом, виданим викладачем. Написати процедуру видалення заданої гілки дерева.

```python
# --- Реалізація бінарного дерева пошуку (BST) та видалення гілки ---

class Node:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None

def insert(root, key):
    """Функція для вставки елементів у дерево"""
    if root is None:
        return Node(key)
    if key < root.key:
        root.left = insert(root.left, key)
    else:
        root.right = insert(root.right, key)
    return root

def delete_branch(root, target_key):
    """Процедура видалення заданої гілки разом з усім піддеревом"""
    if root is None:
        return None
    
    # Якщо лівий або правий син є цільовим — відсікаємо всю гілку
    if root.left and root.left.key == target_key:
        root.left = None
        return root
    if root.right and root.right.key == target_key:
        root.right = None
        return root
    
    # Шукаємо цільовий вузол далі по дереву
    if target_key < root.key:
        root.left = delete_branch(root.left, target_key)
    else:
        root.right = delete_branch(root.right, target_key)
    return root

def print_tree(root, space=0, level_gap=6):
    """Допоміжна функція для наочного відображення дерева"""
    if root is None:
        return
    space += level_gap
    print_tree(root.right, space)
    print(" " * (space - level_gap) + str(root.key))
    print_tree(root.left, space)

# --- ДЕМОНСТРАЦІЯ РОБОТИ З НОВИМИ ЧИСЛАМИ ---

# Новий набір чисел для дерева (корінь тепер 100)
new_keys = [100, 60, 150, 45, 85, 120, 200]
root = None
for k in new_keys:
    root = insert(root, k)

print("--- Початкове бінарне дерево (повернуте на 90° вліво) ---")
print_tree(root)

# Тепер видаляємо іншу гілку — наприклад, 60 (разом з піддеревами 45 та 85)
new_target = 60
root = delete_branch(root, new_target)

print(f"\n--- Дерево після видалення гілки '{new_target}' ---")
print_tree(root)
```

## Оцінка асимптотичної складності операцій з Бінарним Деревом Пошуку 

Ефективність бінарного дерева пошуку критично залежить від його форми (ступеня збалансованості).

| Операція | У середньому (Average case) | В найгіршому разі (Worst case) | Обґрунтування |
| :--- | :---: | :---: | :--- |
| **Search** | $O(\log n)$ | $O(n)$ | У середньому (на збалансованому дереві) ми щокроку відсікаємо половину гілок, тому складність логарифмічна. У найгіршому разі дерево вироджується в лінійний список (наприклад, якщо вставляти числа 1, 2, 3, 4 за зростанням), і пошук перетворюється на лінійний перебір. |
| **Insert** | $O(\log n)$ | $O(n)$ | Щоб вставити елемент, спочатку потрібно знайти його майбутнє місце, що повторює логіку операції пошуку. |
| **Delete** | $O(\log n)$ | $O(n)$ | Класичне видалення вузла або відсікання цілої гілки вимагає її попереднього пошуку в дереві. |

## Написати процедуру генерації купи з будь-якого рандомного масива. Додати до нього елемент, який дорівнює вашому порядковому номеру у списку групи. Вилучити максимальний елемент з купи.

```python
import random

# Функція для проштовхування елемента вгору (при додаванні)
def sift_up(heap, i):
    while i > 1 and heap[i] > heap[i // 2]:
        heap[i], heap[i // 2] = heap[i // 2], heap[i]
        i = i // 2

# Функція для проштовхування елемента вниз (при видаленні)
def sift_down(heap, i):
    n = len(heap) - 1
    while 2 * i <= n:
        left = 2 * i
        right = 2 * i + 1
        largest = left
        
        if right <= n and heap[right] > heap[left]:
            largest = right
            
        if heap[i] >= heap[largest]:
            break
            
        heap[i], heap[largest] = heap[largest], heap[i]
        i = largest

# --- Виконання завдання ---

# 1. Генерація рандомного масиву (7 елементів, як на скриншоті)
random_array = [random.randint(10, 99) for _ in range(7)]
print(f"Згенерований рандомний масив: {random_array}\n")

print("--- Процес генерації купи з масиву ---")
heap = [0] # 0 на початку, як на скриншоті

# Поелементно додаємо числа з масиву в купу, щоб показувати кожен крок
for x in random_array:
    heap.append(x)
    sift_up(heap, len(heap) - 1)
    print(f"Поточний стан купи: {heap}")

# 2. Додавання твого порядкового номера (15)
my_number = 15
print(f"\n--- Додавання вашого номера у групі ({my_number}) ---")
heap.append(my_number)
sift_up(heap, len(heap) - 1)
print(f"Поточний стан купи: {heap}")

# 3. Вилучення максимального елемента
print("\n--- Вилучення максимального елемента ---")
max_val = heap[1] # Корінь купи знаходиться на індексі 1 через нуль попереду
print(f"Вилучено максимальний елемент (корінь): {max_val}")

# Переміщуємо останній елемент на місце кореня і опускаємо його
if len(heap) > 2:
    heap[1] = heap.pop()
    sift_down(heap, 1)
else:
    heap.pop()

print(f"Купа після вилучення максимуму: {heap}")
```

## Оцінка асимптотичної складності операцій з купою (Binary Heap)

Купа — це деревоподібна структура даних, яка зберігається у вигляді масиву та підтримує строгий пріоритет елементів (у Max-Heap корінь завжди є максимальним елементом).

| Операція | У середньому (Average case) | В найгіршому разі (Worst case) | Обґрунтування |
| :--- | :---: | :---: | :--- |
| **Insert** | $O(1)$ | $O(\log n)$ | У середньому новий елемент вставляється внизу і майже не піднімається. У найгіршому разі елемент доведеться просіювати вгору через усю висоту дерева, яка дорівнює $\log_2 n$. |
| **Delete (pop)** | $O(\log n)$ | $O(\log n)$ | Вилучення вершини потребує зміщення останнього елемента на її місце та подальшого його опускання (heapify) вниз по дереву, що завжди займає час, пропорційний висоті дерева. |
| **Search** | $O(n)$ | $O(n)$ | Купа не впорядкована ліворуч-праворуч, як дерево пошуку. Тому для знаходження довільного елемента доведеться виконати повний лінійний перебір масиву. |

## Реалізуйте геш-таблицю з ланцюжковим гешуванням. Проведіть тестування геш-таблиці з різними типами даних (цілі числа, рядки, списки, словники, об’єкти). Перевірити працездатність геш-таблиці з різними типами даних. Виміряти час виконання основних операцій (пошук, вставка, видалення) для різних типів даних. Порівняти результати для різних типів даних.

```python
import time
import pandas as pd

# Клас простісінької Хеш-таблиці з методом ланцюжків для колізій
class HashTable:
    def __init__(self, size=100):
        self.size = size
        self.table = [[] for _ in range(self.size)]
        
    def _get_hash(self, key):
        # Якщо тип даних не хешується (список/словник), трансформуємо його
        if isinstance(key, list):
            key = tuple(key)
        elif isinstance(key, dict):
            key = frozenset(key.items())
        return abs(hash(key)) % self.size

    def insert(self, key, value):
        h = self._get_hash(key)
        for kv in self.table[h]:
            if kv[0] == key:
                kv[1] = value
                return
        self.table[h].append([key, value])

    def search(self, key):
        h = self._get_hash(key)
        for kv in self.table[h]:
            if kv[0] == key:
                return kv[1]
        return None

    def delete(self, key):
        h = self._get_hash(key)
        for i, kv in enumerate(self.table[h]):
            if kv[0] == key:
                del self.table[h][i]
                return True
        return False

# Спеціальний користувацький об'єкт для тестування
class CustomObject:
    def __init__(self, name, val):
        self.name = name
        self.val = val
    def __hash__(self):
        return hash((self.name, self.val))
    def __eq__(self, other):
        return isinstance(other, CustomObject) and self.name == other.name

# --- ТЕСТУВАННЯ ТА ВИМІРЮВАННЯ ЧАСУ ---

# Набір тестових даних різних типів (ключ: значення)
data_types = {
    "Ціле число (int)": (42, "Value_Int"),
    "Рядок (string)": ("kremenchuk", "Value_Str"),
    "Список (list)": ([1, 2, 3], "Value_List"),
    "Словник (dict)": ({"a": 1, "b": 2}, "Value_Dict"),
    "Об'єкт (object)": (CustomObject("test", 99), "Value_Obj")
}

results = []

for name, (key, val) in data_types.items():
    ht = HashTable()
    
    # 1. Замір часу ВСТАВКИ
    t0 = time.perf_counter()
    ht.insert(key, val)
    t1 = time.perf_counter()
    t_insert = t1 - t0
    
    # 2. Замір часу ПОШУКУ
    t0 = time.perf_counter()
    ht.search(key)
    t1 = time.perf_counter()
    t_search = t1 - t0
    
    # 3. Замір часу ВИДАЛЕННЯ
    t0 = time.perf_counter()
    ht.delete(key)
    t1 = time.perf_counter()
    t_delete = t1 - t0
    
    results.append({
        "Тип даних": name,
        "Вставка (сек)": f"{t_insert:.8f}",
        "Пошук (сек)": f"{t_search:.8f}",
        "Видалення (сек)": f"{t_delete:.8f}"
    })

# Виведення результатів у вигляді красивої таблиці DataFrame
df = pd.DataFrame(results)
print("ПОРІВНЯЛЬНИЙ АНАЛІЗ ЧАСУ ВИКОНАННЯ ОПЕРАЦІЙ:")

display(df)
```

### Коротке порівняння результатів для різних типів даних

**Цілі числа (int) та Рядки (str):** Працюють найшвидше. Хеш обчислюється миттєво, а операції виконуються за еталонний константний час $O(1)$.

**Списки (list) та Словники (dict):** Пряме використання неможливе, оскільки вони є змінними (мутабельними). Після обов'язкової конвертації в кортежі (uple) та заморожені множини (frozenset) вони працюють коректно, але трохи повільніше через час на перебір елементів при хешуванні.

**Об'єкти (object):** Працюють стабільно, якщо в класі об'єкта вручну перевизначено метод   _ hash__(). Швидкість операцій залежить від складності полів, з яких зчитується хеш.

**Головний підсумок:** Усі типи даних забезпечують швидкість $O(1)$, але незмінні (імутабельні) типи працюють автоматично, а змінні (мутабельні) потребують обов'язкового перетворення типів.

### Контрольні питання

1. **Бінарне дерево vs Дерево пошуку (BST):** У звичайному бінарному дереві елементи розміщуються хаотично, а в бінарному дереві пошуку (BST) діє чітке правило: лівий нащадок завжди менший за батька, а правий — більший.
2. **Бінарне дерево vs Бінарна купа:** Бінарне дерево не має вимог до значень і форми, тоді як купа обов'язково є повністю заповненим деревом, де кожен батьківський вузол завжди строго більший (або менший) за своїх дітей.
3. **Типи дерев та їх переваги:** Існують прості бінарні дерева, дерева пошуку (BST) для швидкого сортування, збалансовані дерева (AVL, Червоно-чорні) для захисту від втрати швидкодії, та B-дерева, оптимізовані для зберігання великих баз даних.
4. **Застосування дерев:** Дерева ефективно вирішують задачі ієрархічного збереження даних (файлові системи, DOM-модель веб-сторінок), індексації в базах даних та синтаксичного аналізу коду в компиляторах.
5. **Організація та алгоритми купи:** Купа зберігається у вигляді масиву, де при додаванні новий елемент стає в кінець і просіюється вгору (sift_up), а при вилученні кореня його місце займає останній елемент і опускається вниз (sift_down).
6. **Застосування купи:** За допомогою купи ефективно реалізують черги з пріоритетами (планування завдань в ОС), знаходять топ-K значень у великому потоці даних та проводять сортування (Heap Sort).
7. **Принцип роботи геш-функції:** Геш-функція миттєво перетворює унікальний ключ на числовий індекс масиву, що дозволяє записувати та знаходити потрібні дані за цим індексом всього за одну дію ($O(1)$) без перебору всієї таблиці.
