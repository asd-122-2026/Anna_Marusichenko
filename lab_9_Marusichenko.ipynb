## Тестування алгоритмів

```python
def naive_match(p, t):
    """ Наївний (прямий) пошук підрядка p у тексті t """
    assert len(p) <= len(t)
    occurrences = []
    for i in range(0, len(t) - len(p) + 1):
        match = True
        for j in range(0, len(p)):
            if t[i+j] != p[j]:
                match = False
                break
        if match:
            occurrences.append(i)
    return occurrences

def z_func(s):
    """ Обчислення Z-функції для рядка s за лінійний час O(n) """
    Z = [len(s)] + [0] * (len(s) - 1)
    if len(s) <= 1:
        return Z
    
    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            Z[1] += 1
        else:
            break
    
    r, l = 0, 0
    if Z[1] > 0:
        r, l = Z[1], 1
    
    for k in range(2, len(s)):
        if k > r:
            for i in range(k, len(s)):
                if s[i] == s[i - k]:
                    Z[k] += 1
                else:
                    break
            r, l = k + Z[k] - 1, k
        else:
            nbeta = r - k + 1
            Zkp = Z[k - l]
            if nbeta > Zkp:
                Z[k] = Zkp
            else:
                nmatch = 0
                for i in range(r + 1, len(s)):
                    if s[i] == s[i - k]:
                        nmatch += 1
                    else:
                        break
                l, r = k, r + nmatch
                Z[k] = r - k + 1
    return Z

def zMatch(p, t):
    """ Пошук підрядка p в тексті t за допомогою Z-функції """
    s = p + "$" + t
    Z = z_func(s)
    occurrences = []
    for i in range(len(p) + 1, len(s)):
        if Z[i] == len(p):
            occurrences.append(i - (len(p) + 1))
    return occurrences

def compress_with_z(s):
    """ Стиснення періодичного рядка за допомогою Z-функції """
    z_vec = z_func(s)
    for i in range(1, len(s)):
        if (i + z_vec[i] == len(s)) and (z_vec[i] % i == 0):
            return s[:i]
    return s

# =================--- ТЕСТУВАННЯ АЛГОРИТМІВ ---=================

test_text = "banana-ananas-and-panama"
test_pattern = "an"

print("--- Тест 1: Наївний пошук підрядка ---")
print(f"Текст:   '{test_text}'")
print(f"Шаблон:  '{test_pattern}'")
print(f"Індекси входження підрядка: {naive_match(test_pattern, test_text)}")

print("\n--- Тест 2: Обчислення лінійної Z-функції ---")
word = "abcabcabc"
print(f"Рядок:     '{word}'")
print(f"Z-вектор:  {z_func(word)}")

print("\n--- Тест 3: Пошук підрядка через Z-функцію (Z-match) ---")
print(f"Текст:   '{test_text}'")
print(f"Шаблон:  '{test_pattern}'")
print(f"Індекси входження (через Z): {zMatch(test_pattern, test_text)}")

print("\n--- Тест 4: Стиснення періодичного рядка ---")
periodic_str = "xyzxyzxyzxyz"
print(f"Рядок: '{periodic_str}' -> Стиснутий блок: '{compress_with_z(periodic_str)}'")
print(f"Рядок: 'hello'        -> Стиснутий блок: '{compress_with_z('hello')}'")
```

## Асимптотика наївного пошуку

**Формула:** $O(|t| \cdot |p|)$

**Чому:** Зовнішній цикл зміщує шаблон по тексту ($O(|t|)$), а внутрішній у найгіршому випадку перевіряє всі символи шаблону ($O(|p|)$). Найгірший випадок — повторення однакових символів (наприклад, $t = \text{"AAAA"}$, $p = \text{"AA"}$).

## Асимптотика стиснення через Z-функцію

**Формула:** $O(n)$, де $n$ — довжина рядка.

**Чому:** Обчислення Z-функції працює за лінійний час $O(n)$ завдяки скануванню рядка в один бік через «вікно». Фінальний перебір варіантів періоду робить максимум $n$ кроків з операціями за $O(1)$.
