### Функція print_n()

Було реалізовано функцію `print_n()`, яка друкує елементи черги від початку до номера `n` включно.

```python
# --- Реалізація функції для стека ---

def pop_n(stack, n):
    """
    Видаляє елементи стека з його вершини (початку видалення) 
    до номера n включно (вважаємо, що n — це кількість елементів).
    """
    # Перевіряємо, чи є в стеку достатньо елементів
    if n > len(stack):
        print("Помилка: в стеку менше елементів, ніж потрібно видалити!")
        return stack
    
    # Видаляємо n елементів з кінця (вершини стека)
    for _ in range(n):
        stack.pop() # Якщо потрібно видаляти з нульового індексу, пиши pop(0)
        
    return stack

# --- Приклад роботи в Юпітері ---

мій_стек = [10, 20, 30, 40, 50, 60]
print("Початковий стек:", мій_стек)

# Видаляємо 3 елементи з вершини (60, 50, 40)
результат = pop_n(мій_стек, 3)
print("Стек після pop_n(3):", результат)
```

## Оцінка асимптотичної складності операцій зі стеком

Стек є ефективною структурою даних, оскільки всі модифікації (додавання та видалення) відбуваються лише з одного боку — з вершини стека.

| Операція | У середньому (Average case) | В найгіршому разі (Worst case) | Обґрунтування |
| :--- | :---: | :---: | :--- |
| **Insert (push)** | $O(1)$ | $O(1)$ | Елемент завжди додається на вершину стека за одну дію, зсув інших елементів не потрібен. |
| **Delete (pop)** | $O(1)$ | $O(1)$ | Видалення завжди відбувається з вершини стека за одну дію. |
| **Search** | $O(n)$ | $O(n)$ | Для пошуку елемента потрібно послідовно перебрати весь стек. У найгіршому випадку доведеться перевірити всі $n$ елементів. |

## Розглянути самостійно Приклад 3 офіційної документації модуля Queue [31];

```python
import queue
import threading
import time

q = queue.Queue()

def worker():
    while True:
        item = q.get()
        print(f"Оброблено: {item}")
        time.sleep(1)

threading.Thread(target=worker, daemon=True).start()

for i in range(5):
    q.put(i)
    time.sleep(0.5)

print("Готово.")
```

## Написати функцію print_n(), що друкує елементи черги з його початку до номера n включно;

```python
from collections import deque

def print_n(queue, n):

    if not queue:
        print("Черга порожня")
        return

    print(f"Перші {n} елементів:")
    
    for i, element in enumerate(queue):
        if i < n:
            print(f"{i + 1}: {element}")
        else:
            break

my_queue = deque(['Яблуко', 'Банан', 'Вишня', 'Драгонфрут', 'Екзотик'])

print_n(my_queue, 3)
```

## Оцінка асимптотичної складності операцій з чергою (Queue)

Черга є базовою структурою даних, яка функціонує за принципом FIFO (First In, First Out). Взаємодія з чергою відбувається з двох кінців: додавання в кінець (tail), а видалення — з початку (head).

| Операція | У середньому (Average case) | В найгіршому разі (Worst case) | Обґрунтування |
| :--- | :---: | :---: | :--- |
| **Insert (enqueue)** | $O(1)$ | $O(1)$ | Новий елемент завжди вставляється в кінець черги за одну фіксовану дію. |
| **Delete (dequeue)** | $O(1)$ | $O(1)$ | Видалення елемента завжди відбувається з початку черги, що також займає один крок. |
| **Search** | $O(n)$ | $O(n)$ | Щоб знайти елемент всередині черги, її потрібно послідовно перебрати. У найгіршому разі елемент виявиться останнім або його взагалі не буде, що потребує $n$ кроків. |

# Контрольні питання

### 1. Що таке стек і які операції можна виконувати зі стеком?
Стек — це структура даних, що працює за принципом «останнім прийшов, першим пішов» (LIFO), і дозволяє додавати елементи на вершину (push), видаляти їх звідти (pop), переглядати верхнє значення (peek) та перевіряти структуру на порожнечу (isEmpty).
### 2. Яка основна відмінність між стеком та чергою?
Основна відмінність полягає в порядку обробки даних: у стеку першим видаляється елемент, який додали найостаннішим (LIFO), а в черзі — той, який додали найпершим (FIFO).
### 3. Як ви можете реалізувати стек за допомогою масиву і за допомогою зв’язаного списку? Які переваги та недоліки кожного підходу?
Стек на масиві реалізується у фіксованій пам'яті за допомогою індексу вершини, що працює швидко, але обмежує розмір, тоді як стек на зв'язаному списку динамічно додає вузли через вказівники, що дозволяє структурі нескінченно рости, але витрачає більше пам'яті.
### 4. Які є застосування стека та черги в програмуванні і реальному житті?
У програмуванні та житті стек використовується для скасування дій (Ctrl+Z) і зберігання стопок тарілок, а черга — для планування завдань ОС, друку на принтері та обслуговування покупців у супермаркетах.
