# Рукурсія

**Рекурсія** — це спосіб організації алгоритму, при якому функція викликає сама себе для розв'язання підзадачі меншого розміру.

Для правильної роботи рекурсивної функції необхідно визначити:

- базовий випадок (умову завершення рекурсії);
- рекурсивний виклик.

Одним із поширених підходів до побудови ефективних алгоритмів є стратегія **«розділяй і володарюй»**.

Її суть полягає в тому, що задача:

1. Розбивається на менші підзадачі.
2. Кожна підзадача розв'язується окремо.
3. Отримані результати об'єднуються.

До алгоритмів, що використовують цей підхід, належить алгоритм **Merge Sort (сортування злиттям)**.

### Оцінка асимптотичної складності сортування злиттям

За Головною теоремою рекурсії час роботи алгоритму записується формулою:

#### T(n) = 2T(n/2) + O(n)

* **$2T(n/2)$** — масив на кожному кроці ділиться навпіл. Оскільки ділимо завжди на 2, кількість таких рівнів розбиття дорівнює $\log_2 n$.
* **$O(n)$** — це час на злиття шматочків докупи (функція `merge`). Вона просто по черзі перебирає всі елементи.

Якщо об'єднати ці дві дії, загальна часова складність алгоритму буде:

#### O(n \log n)

Ця складність однакова і для найкращого, і для найгіршого випадків, тому що алгоритм завжди ділить масив і зливає його за однією схемою.

**Складність за пам'яттю:** становить **$O(n)$**. Алгоритм не може сортувати елементи всередині початкового масиву, тому функція `merge` завжди створює нові додаткові масиви в пам'яті для копіювання елементів.

### Числа Фібоначчі

```python
def fibonacci(n):

    if n == 0:
        return 0

    if n == 1 or n == 2:
        return 1

    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(10))
```

### Факторіал через while

```python
def fac_while(n):
    factorial = 1

    while n > 0:
        factorial *= n
        n -= 1

    return factorial

print(fac_while(5))
```

### Рекурсивний факторіал

```python
def fac(n):

    if n == 0:
        return 1

    return fac(n - 1) * n

print(fac(5))
```

### Сортування злиттям

```python
def merge(left, right):

    result = []

    i = 0
    j = 0

    while i < len(left) and j < len(right):

        if left[i] <= right[j]:
            result.append(left[i])
            i += 1

        else:
            result.append(right[j])
            j += 1

    result += left[i:]
    result += right[j:]

    return result


def mergesort(lst):

    if len(lst) < 2:
        return lst

    middle = len(lst) // 2

    left = mergesort(lst[:middle])
    right = mergesort(lst[middle:])

    return merge(left, right)


a = [6, 5, 3, 1, 8, 7, 2, 4]

print(mergesort(a))
```
